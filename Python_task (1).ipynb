Prg-2:Factorial of a number

Prg-3:Prime Number or not

```python
n =34

if n < 2:
    print("Not a prime number")
else:
    is_prime = True

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            is_prime = False
            break

    if is_prime:
        print("Prime number")
    else:
        print("Not a prime number")
```

Prg:6 Character frequency

```python
s = "Pineapple"
freq = {}
for ch in s:
    freq[ch] = freq.get(ch, 0) + 1
print("Character frequency:", freq)
```

Prg:7 Armstrong Number or not

```python
n = 153
total = sum(int(d) ** len(str(n)) for d in str(n))
print("Armstrong:", total == n)
```

Prg-8:Longest substring without repeated characters

```python
s = "abcabcbb"

seen = {}
left = 0
longest = 0

for right in range(len(s)):
    if s[right] in seen and seen[s[right]] >= left:
        left = seen[s[right]] + 1

    seen[s[right]] = right
    longest = max(longest, right - left + 1)

print(longest)
```

Prg-9: Group Anagrams

```python
words = ["eat", "tea", "tan", "ate", "nat", "bat"]

groups = {}

for word in words:
    key = "".join(sorted(word))
    if key not in groups:
        groups[key] = []
    groups[key].append(word)

print(list(groups.values()))
```

Prg-10:Maximum sub array

```python
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

cur = nums[0]
best = nums[0]

for i in range(1, len(nums)):
    cur = max(nums[i], cur + nums[i])
    best = max(best, cur)

print(best)
```

**SQL QUERIES Task**





```python
import sqlite3

con = sqlite3.connect(":memory:")
cur = con.cursor()

cur.execute("""
CREATE TABLE Student (
    id INTEGER,
    name TEXT,
    dept TEXT,
    salary INTEGER,
    manager_id INTEGER
)
""")

cur.executemany("""
INSERT INTO Student VALUES (?, ?, ?, ?, ?)
""", [
    (1, "Ravi", "CSE", 60000, None),
    (2, "Gopi", "CSE", 75000, 1),
    (3, "Rani", "ECE", 50000, None),
    (4, "Sita", "ECE", 55000, 3),
    (5, "Kiran", "CSE", 75000, 1),
    (6, "Arun", "EEE", 45000, None)
])

def show(title, query):
    print("\n" + title)
    cur.execute(query)
    print([col[0] for col in cur.description])
    for row in cur.fetchall():
        print(row)

show("1. Display All Records",
     "SELECT * FROM Student")

show("2. Salary Greater Than 55000",
     "SELECT name, salary FROM Student WHERE salary > 55000")

show("3. Sort by Salary",
     "SELECT name, salary FROM Student ORDER BY salary DESC")

show("4. Count Students by Department",
     "SELECT dept, COUNT(*) AS total FROM Student GROUP BY dept")

show("5. Average Salary by Department",
     "SELECT dept, AVG(salary) AS average FROM Student GROUP BY dept")

show("6. Departments with More Than One Record",
     "SELECT dept, COUNT(*) AS total FROM Student GROUP BY dept HAVING COUNT(*) > 1")

show("7. Highest Salary",
     "SELECT MAX(salary) AS highest_salary FROM Student")

show("8. Second Highest Distinct Salary",
     "SELECT MAX(salary) AS second_highest FROM Student WHERE salary < (SELECT MAX(salary) FROM Student)")

show("9. Employees Earning the Highest Salary",
     "SELECT name, salary FROM Student WHERE salary = (SELECT MAX(salary) FROM Student)")

show("10. Employee and Manager Names",
     """SELECT e.name AS employee, m.name AS manager
        FROM Student e
        JOIN Student m ON e.manager_id = m.id""")

con.close()
```

QUERIES USING JOINS

```python
import sqlite3

con = sqlite3.connect(":memory:")
cur = con.cursor()

cur.execute("CREATE TABLE Student (id INTEGER, name TEXT, dept_id INTEGER, marks INTEGER)")
cur.execute("CREATE TABLE Dept (dept_id INTEGER, dept_name TEXT)")
cur.execute("CREATE TABLE Course (course_id INTEGER, course_name TEXT)")
cur.execute("CREATE TABLE Enroll (student_id INTEGER, course_id INTEGER)")

cur.executemany("INSERT INTO Student VALUES (?, ?, ?, ?)", [
    (1, "Ravi", 10, 85),
    (2, "Gopi", 20, 92),
    (3, "Sita", 10, 78),
    (4, "Arun", None, 65),
    (5, "Rani", 30, 88)
])

cur.executemany("INSERT INTO Dept VALUES (?, ?)", [
    (10, "CSE"),
    (20, "ECE"),
    (40, "EEE")
])

cur.executemany("INSERT INTO Course VALUES (?, ?)", [
    (1, "Python"),
    (2, "SQL"),
    (3, "Java")
])

cur.executemany("INSERT INTO Enroll VALUES (?, ?)", [
    (1, 1),
    (1, 2),
    (2, 2),
    (3, 1),
    (5, 3)
])

def show(title, query):
    print("\n" + title)
    cur.execute(query)
    print([col[0] for col in cur.description])
    for row in cur.fetchall():
        print(row)

show("1. INNER JOIN",
     """SELECT s.name, d.dept_name
        FROM Student s
        INNER JOIN Dept d ON s.dept_id = d.dept_id""")

show("2. LEFT JOIN",
     """SELECT s.name, d.dept_name
        FROM Student s
        LEFT JOIN Dept d ON s.dept_id = d.dept_id""")

show("3. Students Without a Matching Department",
     """SELECT s.name
        FROM Student s
        LEFT JOIN Dept d ON s.dept_id = d.dept_id
        WHERE d.dept_id IS NULL""")

show("4. Students with Their Courses",
     """SELECT s.name, c.course_name
        FROM Student s
        JOIN Enroll e ON s.id = e.student_id
        JOIN Course c ON e.course_id = c.course_id""")

show("5. Count Students in Each Department",
     """SELECT d.dept_name, COUNT(s.id) AS total
        FROM Dept d
        LEFT JOIN Student s ON d.dept_id = s.dept_id
        GROUP BY d.dept_id, d.dept_name""")

show("6. Students Scoring Above Average",
     """SELECT name, marks
        FROM Student
        WHERE marks > (SELECT AVG(marks) FROM Student)""")

show("7. Highest Marks in Each Department",
     """SELECT d.dept_name, MAX(s.marks) AS highest_marks
        FROM Student s
        JOIN Dept d ON s.dept_id = d.dept_id
        GROUP BY d.dept_id, d.dept_name""")

show("8. Classify Students by Marks",
     """SELECT name, marks,
        CASE
            WHEN marks >= 90 THEN 'Excellent'
            WHEN marks >= 75 THEN 'Good'
            ELSE 'Average'
        END AS grade
        FROM Student""")

show("9. Rank Students by Marks",
     """SELECT name, marks,
        RANK() OVER (ORDER BY marks DESC) AS student_rank
        FROM Student""")

show("10. Count Courses per Student",
     """SELECT s.name, COUNT(e.course_id) AS total_courses
        FROM Student s
        LEFT JOIN Enroll e ON s.id = e.student_id
        GROUP BY s.id, s.name""")

show("11. CROSS JOIN",
     """SELECT s.name, c.course_name
        FROM Student s
        CROSS JOIN Course c""")

show("12. Second Highest Distinct Marks",
     """SELECT MAX(marks) AS second_highest
        FROM Student
        WHERE marks < (SELECT MAX(marks) FROM Student)""")

con.close()
```
